```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
Vref = 3.3
Vdd = 5
```

# Canal R

```python
df_R = pd.read_json("measurements_R.json").set_index("index")
df_R = df_R.drop("data", axis="columns")
df_R = df_R.rename(columns={"mean": "adc"})
df_R.tail()
```

```python
df_R["Vo"] = df_R["adc"] * Vref / 4095 
df_R["G"] = (Vref - df_R["Vo"]) / (df_R["Rf"]*(Vdd-Vref))
df_R.tail()
```

```python
df_R1 = pd.read_json("measurements_R_1.json").set_index("index")
df_R1 = df_R1.drop("data", axis="columns")
df_R1 = df_R1.rename(columns={"mean": "adc"})
df_R1.tail()
```

```python
df_R1["Vo"] = df_R1["adc"] * Vref / 4095 
df_R1["G"] = (Vref - df_R1["Vo"]) / (df_R1["Rf"]*(Vdd-Vref))
df_R1.tail()
```

```python
plt.semilogx("intensity", "G", data=df_R, marker=".", label="muestra 1", alpha=0.6)
plt.semilogx("intensity", "G", data=df_R1, marker=".", label="muestra 2", alpha=0.6)
plt.legend()
plt.grid()
```
