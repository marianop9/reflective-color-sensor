```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# plots de la respuesta ideal de un LDR, con eje y logaritmico y lineal.
# usa una ganancia de 100

E = np.logspace(-2, 0.5, 100)

fig, (ax1, ax2) = plt.subplots(2, 1)

for gamma in [0.7,0.8,0.9]:
    Rldr = 100 * np.pow(E, -gamma)
    ax1.loglog(E, Rldr, label=f"g={gamma:.1}")
    ax2.semilogx(E, Rldr)

ax1.legend()
fig.suptitle("Respuesta ideal del LDR (R)")
fig.supxlabel("$E$ (iluminancia) [lux]")
_ = fig.supylabel(r"$R$ [$\Omega$]")
```

```python
# plots de la respuesta ideal de conductancia de un LDR, con eje y logaritmico y lineal.
# usa una ganancia de 100

E = np.logspace(1.5, 4, 100)

fig, (ax1, ax2) = plt.subplots(2, 1)

for gamma in [0.7,0.8,0.9]:
    Gldr = 1/100 * np.pow(E, gamma)
    ax1.loglog(E, Gldr, label=f"g={gamma:.1}")
    ax2.semilogx(E, Gldr)

ax1.legend()
fig.suptitle("Respuesta ideal del LDR (G)")
fig.supxlabel("$E$ (iluminancia) [lux]")
_ = fig.supylabel(r"$G$ [$1/\Omega$]")
```
