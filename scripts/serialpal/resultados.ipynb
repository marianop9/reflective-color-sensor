```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def show_color_comparison(measured, reference, title=""):
    # Create 1x2 image: left = measured, right = reference
    img = np.zeros((100, 200, 3), dtype=np.uint8)
    img[:, :100, :] = measured
    img[:, 100:, :] = reference

    plt.figure()
    plt.imshow(img)
    plt.axis("off")
    plt.title(title)
    plt.show()
```

```python
medido = (159, 255, 179)
ref = (137, 162, 84)

show_color_comparison(  
    medido,
    ref,
    "Verde"
)
```

```python
medido = (239, 255, 218)
ref = (187, 183, 104)

show_color_comparison(  
    medido,
    ref,
    "Amarillo"
)
```

```python
medido = (134, 255, 255)
ref = (123, 171, 179),
show_color_comparison(  
    medido, 
    ref,
    "Celeste"
)
```

```python
medido = (229, 177, 211)
# ref = (202, 129, 133),
ref = (214, 140, 142)

show_color_comparison(  
    medido, 
    ref,
    "Rosa"
)
```
