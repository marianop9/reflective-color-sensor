Los valores de referencia se obtuvieron utilizando la aplicación *Color Picker - Pixel Color* en un iPhone 16.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# def show_color_comparison(measured, reference, title=""):
#     # Create 1x2 image: left = measured, right = reference
#     img = np.zeros((100, 200, 3), dtype=np.uint8)
#     img[:, :100, :] = measured
#     img[:, 100:, :] = reference

#     plt.figure()
#     plt.imshow(img)
#     plt.axis("off")
#     plt.title(title)
#     plt.show()

def show_color_comparison(measured, reference, title=""):
    measured = np.array(measured, dtype=np.uint8)
    reference = np.array(reference, dtype=np.uint8)

    # Create side-by-side image
    img = np.zeros((100, 200, 3), dtype=np.uint8)
    img[:, :100, :] = measured
    img[:, 100:, :] = reference

    def get_text_color(rgb):
        # Perceived luminance (simple approximation)
        luminance = 0.299*rgb[0] + 0.587*rgb[1] + 0.114*rgb[2]
        return "black" if luminance > 186 else "white"

    fig, ax = plt.subplots()
    ax.imshow(img)
    ax.axis("off")

    # Add RGB labels
    ax.text(
        50, 50,
        f"Medido\nrgb({", ".join(str(x) for x in measured)})",
        ha="center", va="center",
        color=get_text_color(measured),
        fontsize=10, weight="bold"
    )

    ax.text(
        150, 50,
        f"Referencia\nrgb({", ".join(str(x) for x in reference)})",
        ha="center", va="center",
        color=get_text_color(reference),
        fontsize=10, weight="bold"
    )

    if title:
        ax.set_title(title)

    plt.tight_layout()
    plt.show()
```

```python
medido = (159, 255, 179)
ref = (137, 162, 84)

show_color_comparison(  
    medido,
    ref,
    "Verde"
)
```

```python
medido = (239, 255, 218)
ref = (187, 183, 104)

show_color_comparison(  
    medido,
    ref,
    "Amarillo"
)
```
